### CPT E/M codes

```python
# Define valid codes
new_patient_codes = {
    "99201",
    "99202",
    "99203",
    "99204",
    "99205",
    "99381",
    "99382",
    "99383",
    "99384",
    "99385",
    "99386",
    "99387",
    "99241",
    "99242",
    "99243",
    "99244",
    "99245",
}

established_patient_codes = {
    "99211",
    "99212",
    "99213",
    "99214",
    "99215",
    "99391",
    "99392",
    "99393",
    "99394",
    "99395",
    "99396",
    "99397",
    "99251",
    "99252",
    "99253",
    "99254",
    "99255",
}

valid_em_codes = new_patient_codes.union(established_patient_codes)
```

```python
from tqdm import tqdm
from collections import Counter


def get_cpt_code_statistics(charge_table_path) -> Counter:
    code_counter = Counter()

    charges_df = pd.read_csv(charge_table_path)
    parent_charge_ids = charges_df["hash_PARENT_CHARGE_ITEM_ID"].values
    final = charges_df[~charges_df["hash_CHARGE_ITEM_ID"].isin(parent_charge_ids)]

    for eid, group_df in tqdm(final.groupby("hash_ENCNTR_ID")):
        try:
            coder_codes = set()
            for _, row in group_df.iterrows():
                code = str(row["FIELD6"]).strip()
                if code in valid_em_codes:
                    coder_codes.add(code)

            for code in coder_codes:
                code_counter[code] += 1
        except:
            print("Unable to extract code for ", eid)

    return code_counter
```

```python
hudson_cpt_code_counter = get_cpt_code_statistics(hudson_charges)
```

```python
uni_missouri_cpt_code_counter = get_cpt_code_statistics(uni_missouri_charges)
```

```python
st_john_cpt_code_counter = get_cpt_code_statistics(st_john_charges)
```

```python
combined_cpt_code_counter = hudson_cpt_code_counter + uni_missouri_cpt_code_counter + st_john_cpt_code_counter
```

```python
counter_dict = {
    "hudson": hudson_cpt_code_counter,
    "uni_missouri": uni_missouri_cpt_code_counter,
    "st_johns": st_john_cpt_code_counter,
    "all": combined_cpt_code_counter
}

json.dump(counter_dict, open("em_codes.json", "w+"), indent=4)
```

```python
len(counter_dict["all"])
```

```python
for i, freq in combined_cpt_code_counter.most_common(10):
    print(i, code_lookup[i], freq)
```

### CPT Procedure codes

```python
def get_cpt_procedure_statistics(charge_table_path) -> Counter:
    code_counter = Counter()

    charges_df = pd.read_csv(charge_table_path)
    parent_charge_ids = charges_df["hash_PARENT_CHARGE_ITEM_ID"].values
    final = charges_df[~charges_df["hash_CHARGE_ITEM_ID"].isin(parent_charge_ids)]

    for eid, group_df in tqdm(final.groupby("hash_ENCNTR_ID")):

        coder_codes = set()
        group_df = group_df.loc[group_df["EXP"].str.contains("CPT", na=False)]
        for _, row in group_df.iterrows():
            code = str(row["FIELD6"]).strip()
            if code not in valid_em_codes and code.isnumeric() and len(code) == 5:
                coder_codes.add(code)

        for code in coder_codes:
            code_counter[code] += 1


    return code_counter
```

```python
uni_missouri_procedure_counter = get_cpt_procedure_statistics(uni_missouri_charges)
```

```python
st_johns_procedure_counter = get_cpt_procedure_statistics(st_john_charges)
```

```python
combined_procedure_counter = hudson_procedure_counter + uni_missouri_procedure_counter + st_johns_procedure_counter
```
