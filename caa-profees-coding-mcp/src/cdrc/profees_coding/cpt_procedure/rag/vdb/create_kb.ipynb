```python
import json
import pandas as pd

kb_path = "CPT long descriptions.csv"
kb_df = pd.read_csv(kb_path)

code_list = []
for idx, row in kb_df.iterrows():
    code = {"code": row["CPT_Code"], "description": row["CPT_Long_Description"]}
    code_list.append(code)

with open("cpt_long_descriptions.json", "w") as f:
    json.dump(code_list, f, indent=2)
```
